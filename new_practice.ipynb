```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
test={
    'science':[99,93,89,78],
    'social studies':[89,88,98,84],
    'math':[87,77,90,67]
}
sub=pd.DataFrame(test)
sub
```

```python
sub.sum(axis=0)
```

```python
sub.rename_axis(index='RollNo.',inplace=True)
```

```python
sub.reset_index(inplace=True)
```

```python
sub['RollNo.']=sub['RollNo.'].shift(-1)
```

```python
sub['RollNo.']=np.where(sub['RollNo.'].isna(),sub['RollNo.'].ffill()+1,sub['RollNo.'])
```

```python
sub
```

```python
sub['RollNo.']=sub['RollNo.'].astype('int32')
```

```python
sub
```

```python
sub_melt=pd.melt(sub,id_vars='RollNo.',var_name='Subjects',value_name='Score')
sub_melt
```

```python
sub_melt['Score'].values
```

```python
fig=px.line(sub_melt,x=sub_melt['RollNo.'],y=sub_melt['Score'],markers=True,color=sub_melt['Subjects'])
fig.update_layout(
    xaxis=dict(type='category'),
    title_text="Score by Roll Number and Subject",
    xaxis_title="Roll Number",
    yaxis_title="Score"
)
fig.show()
```

```python
fig2=px.bar(sub_melt,x='RollNo.',y='Score',color='Subjects',barmode='group',
                color_discrete_sequence=px.colors.qualitative.Pastel2)
fig2.update_layout(
    xaxis=dict(type='category'),
    title_text="Score by Roll Number and Subject",
    xaxis_title="Roll Number",
    yaxis_title="Score"
)
fig2.show()
```
