```python
from pyxlsb import open_workbook
import pandas as pd

def convert_xlsb_to_xlsx(xlsb_path,xlsx_path):
    with open_workbook(xlsb_path) as wb:
        with wb.get_sheet(1) as sheet:
            data=[[item.v for item in row] for row in sheet.rows()]

    df=pd.DataFrame(data[1:],columns=data[0])

    df.to_excel(xlsx_path,index=False)

convert_xlsb_to_xlsx(r'C:\Users\Dell\Downloads\Norm DJSJ.xlsb',r'C:\Users\Dell\Documents\Norm.xlsx')
```

```python
import pandas as pd
import numpy as np

def straight_line_amortization(principal, annual_rate, years):
    """
    Create a straight-line amortization schedule
    
    Args:
    principal (float): Total loan amount
    annual_rate (float): Annual interest rate
    years (int): Loan tenure in years
    
    Returns:
    pandas.DataFrame: Amortization schedule
    """
    # Calculations
    annual_principal = principal / years
    periods = years
    
    # Initialize lists to store schedule data
    payment_num = []
    begin_balance = []
    principal_payment = []
    interest_payment = []
    total_payment = []
    end_balance = []
    
    # Current balance starts at principal
    current_balance = principal
    
    for period in range(1, periods + 1):
        # Interest is calculated on remaining balance
        interest = current_balance * annual_rate
        
        # Principal payment is fixed
        current_principal = annual_principal
        
        # Total payment
        current_total = current_principal + interest
        
        # Track data
        payment_num.append(period)
        begin_balance.append(current_balance)
        principal_payment.append(current_principal)
        interest_payment.append(interest)
        total_payment.append(current_total)
        
        # Reduce balance
        current_balance -= current_principal
        end_balance.append(current_balance)
    
    # Create DataFrame
    df = pd.DataFrame({
        'Payment Number': payment_num,
        'Beginning Balance': begin_balance,
        'Principal Payment': principal_payment,
        'Interest Payment': interest_payment,
        'Total Payment': total_payment,
        'Ending Balance': end_balance
    })
    
    # Format numeric columns
    numeric_cols = ['Beginning Balance', 'Principal Payment', 
                    'Interest Payment', 'Total Payment', 'Ending Balance']
    df[numeric_cols] = df[numeric_cols].round(2)
    
    return df

# Generate Straight-Line Amortization Schedule
straight_schedule = straight_line_amortization(
    principal=6000000, 
    annual_rate=0.24, 
    years=7
)

print("Straight-Line Amortization Schedule:")
print(straight_schedule.to_string(index=False))

# Save to CSV for reference
straight_schedule.to_csv('straight_line_amortization.csv', index=False)
```

```python
import pandas as pd
import numpy as np

def annuity_amortization(principal, annual_rate, years):
    """
    Create an annuity amortization schedule
    
    Args:
    principal (float): Total loan amount
    annual_rate (float): Annual interest rate
    years (int): Loan tenure in years
    
    Returns:
    pandas.DataFrame: Amortization schedule
    """
    # Calculations
    periods = years
    monthly_rate = annual_rate / 12
    
    # Calculate fixed monthly payment
    monthly_payment = principal * (monthly_rate * (1 + monthly_rate)**(periods*12)) / ((1 + monthly_rate)**(periods*12) - 1)
    
    # Initialize lists to store schedule data
    payment_num = []
    begin_balance = []
    principal_payment = []
    interest_payment = []
    total_payment = []
    end_balance = []
    
    # Current balance starts at principal
    current_balance = principal
    
    for period in range(1, periods * 12 + 1):
        # Interest for the period
        interest = current_balance * monthly_rate
        
        # Principal payment is total payment minus interest
        current_principal = monthly_payment - interest
        
        # Track data
        payment_num.append(period)
        begin_balance.append(current_balance)
        principal_payment.append(current_principal)
        interest_payment.append(interest)
        total_payment.append(monthly_payment)
        
        # Reduce balance
        current_balance -= current_principal
        end_balance.append(current_balance)
    
    # Create DataFrame
    df = pd.DataFrame({
        'Payment Number': payment_num,
        'Beginning Balance': begin_balance,
        'Principal Payment': principal_payment,
        'Interest Payment': interest_payment,
        'Total Payment': total_payment,
        'Ending Balance': end_balance
    })
    
    # Format numeric columns
    numeric_cols = ['Beginning Balance', 'Principal Payment', 
                    'Interest Payment', 'Total Payment', 'Ending Balance']
    df[numeric_cols] = df[numeric_cols].round(2)
    
    return df

# Generate Annuity Amortization Schedule
annuity_schedule = annuity_amortization(
    principal=6000000, 
    annual_rate=0.24, 
    years=7
)

print("Annuity Amortization Schedule (First and Last 5 Payments):")
# Display first and last 5 payments
display_schedule = pd.concat([
    annuity_schedule.head(), 
    annuity_schedule.tail()
])
print(display_schedule.to_string(index=False))

# Save to CSV for reference
annuity_schedule.to_csv('annuity_amortization.csv', index=False)

# Additional Summary
print("\nSummary:")
print(f"Total Payments: {len(annuity_schedule)}")
print(f"Monthly Payment: ${annuity_schedule['Total Payment'].iloc[0]:.2f}")
print(f"Total Interest Paid: ${annuity_schedule['Interest Payment'].sum():.2f}")
```

```python
def calculate_emi(principal, annual_rate, years):
    # Monthly interest rate
    monthly_rate = annual_rate / 12 / 100
    
    # Total number of months
    num_months = years * 12
    
    # EMI Calculation
    emi = principal * monthly_rate * (1 + monthly_rate)**num_months / ((1 + monthly_rate)**num_months - 1)
    
    # Total amount payable
    total_payable = emi * num_months
    
    # Total interest
    total_interest = total_payable - principal
    
    return {
        'Monthly EMI': round(emi, 2),
        'Total Payable': round(total_payable, 2),
        'Total Interest': round(total_interest, 2)
    }

# Calculate EMI
result = calculate_emi(165000, 11.4, 3)
print("EMI Calculation Details:")
for key, value in result.items():
    print(f"{key}: ₹{value:,.2f}")
```

```python
import pandas as pd
import numpy as np

def generate_amortization_schedule(principal, annual_rate, years):
    # Monthly calculations
    monthly_rate = annual_rate / 12 / 100
    num_months = years * 12
    
    # Calculate EMI
    emi = principal * monthly_rate * (1 + monthly_rate)**num_months / ((1 + monthly_rate)**num_months - 1)
    
    # Initialize lists to store schedule data
    schedule = []
    remaining_balance = principal
    
    for month in range(1, num_months + 1):
        # Calculate interest for the month
        interest_payment = remaining_balance * monthly_rate
        
        # Calculate principal payment
        principal_payment = emi - interest_payment
        
        # Update remaining balance
        remaining_balance -= principal_payment
        
        # Store monthly details
        schedule.append({
            'Month': month,
            'EMI': emi,
            'Principal Payment': principal_payment,
            'Interest Payment': interest_payment,
            'Remaining Balance': max(0, remaining_balance)
        })
    
    # Create DataFrame
    df = pd.DataFrame(schedule)
    
    # Round numeric columns
    numeric_cols = ['EMI', 'Principal Payment', 'Interest Payment', 'Remaining Balance']
    df[numeric_cols] = df[numeric_cols].round(2)
    
    return df

# Generate full amortization schedule
amortization_df = generate_amortization_schedule(400000, 11, 4)

# Print selected portions of the schedule
print("First 5 Months:")
print(amortization_df.head().to_string(index=False))

print("\nLast 5 Months:")
print(amortization_df.tail().to_string(index=False))

print("\nSummary:")
print(f"Total Interest Paid: ₹{amortization_df['Interest Payment'].sum():.2f}")
print(f"Total Principal Paid: ₹{amortization_df['Principal Payment'].sum():.2f}")
print(f"Monthly EMI: ₹{amortization_df['EMI'].iloc[0]:.2f}")

# Optional: Save full schedule to CSV
amortization_df.to_csv('loan_amortization_schedule.csv', index=False)
```

```python
import pandas as pd
import numpy as np

def generate_amortization_schedule(principal, annual_rate, years):
    # Monthly calculations
    monthly_rate = annual_rate / 12 / 100
    num_months = years * 12
    
    # Calculate EMI
    emi = principal * monthly_rate * (1 + monthly_rate)**num_months / ((1 + monthly_rate)**num_months - 1)
    
    # Initialize lists to store schedule data
    schedule = []
    remaining_balance = principal
    
    for month in range(1, num_months + 1):
        # Calculate interest for the month
        interest_payment = remaining_balance * monthly_rate
        
        # Calculate principal payment
        principal_payment = emi - interest_payment
        
        # Update remaining balance
        remaining_balance -= principal_payment
        
        # Store monthly details
        schedule.append({
            'Month': month,
            'EMI': emi,
            'Principal Payment': principal_payment,
            'Interest Payment': interest_payment,
            'Remaining Balance': max(0, remaining_balance)
        })
    
    # Create DataFrame
    df = pd.DataFrame(schedule)
    
    # Round numeric columns
    numeric_cols = ['EMI', 'Principal Payment', 'Interest Payment', 'Remaining Balance']
    df[numeric_cols] = df[numeric_cols].round(2)
    
    return df

def calculate_foreclosure(df, total_months, foreclosure_fee_percentage=0.02):
    # Get the last month's balance (37th month)
    if total_months + 1 > len(df):
        remaining_balance = df['Remaining Balance'].iloc[-1]
    else:
        remaining_balance = df['Remaining Balance'].iloc[total_months]
    
    # Calculate foreclosure fee
    foreclosure_fee = remaining_balance * foreclosure_fee_percentage
    
    # Total foreclosure amount
    total_foreclosure_amount = remaining_balance + foreclosure_fee
    
    # Total interest paid until foreclosure
    total_interest_paid = df.loc[df['Month'] <= total_months, 'Interest Payment'].sum()
    
    return {
        'Remaining Balance': remaining_balance,
        'Foreclosure Fee (2.85%)': foreclosure_fee,
        'Total Foreclosure Amount': total_foreclosure_amount,
        'Total Interest Paid Until Foreclosure': total_interest_paid
    }

# Generate full amortization schedule
amortization_df = generate_amortization_schedule(400000, 11, 4)

# Perform foreclosure calculation
foreclosure_details = calculate_foreclosure(amortization_df, 37)

# Total amount paid by foreclosing
emi_paid_till_foreclosure = amortization_df.loc[amortization_df['Month'] <= 37, 'EMI'].sum()
total_foreclosure_paid = emi_paid_till_foreclosure + foreclosure_details['Total Foreclosure Amount']

# Compare with continuing loan
total_payments = amortization_df['EMI'].sum()
difference = total_payments - total_foreclosure_paid

# Display results
print("Foreclosure Details:")
for key, value in foreclosure_details.items():
    print(f"{key}: ₹{value:,.2f}")

print(f"\nTotal Payments if Continuing Loan: ₹{total_payments:,.2f}")
print(f"Total Amount Paid by Foreclosing: ₹{total_foreclosure_paid:,.2f}")
print(f"Difference Between Continuing Loan and Foreclosing: ₹{difference:,.2f}")
```
